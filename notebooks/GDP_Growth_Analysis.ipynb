```python
# 1. تثبيت واستدعاء المكتبات الأساسية لتحليل البيانات
import pandas as pd
import requests
import plotly.express as px

# 2. تحديد الدول والمؤشرات الاقتصادية المراد سحبها من البنك الدولي
# NY.GDP.MKTP.KD.ZG = معدل نمو الناتج المحلي الإجمالي السنوي (%)
# FP.CPI.TOTL.ZG = معدل التضخم السنوي (%)
countries = ['IRQ', 'SAU', 'TUR', 'RUS', 'DEU']  # العراق، السعودية، تركيا، روسيا، ألمانيا
indicator = 'NY.GDP.MKTP.KD.ZG'

# 3. جلب البيانات عبر API البنك الدولي للفترة من 2015 إلى 2025
url = f"http://api.worldbank.org/v2/country/{';'.join(countries)}/indicator/{indicator}?date=2015:2025&format=json&per_page=1000"
response = requests.get(url).json()

# 4. تحويل البيانات إلى الجدول الإحصائي (DataFrame)
data = response[1]
df = pd.DataFrame([{
    'Country': item['country']['value'],
    'Year': int(item['date']),
    'GDP_Growth': item['value']
} for item in data if item['value'] is not None])

# 5. عرض أول 5 أسطر من البيانات المترجمة للجدول
print("--- تم جلب البيانات بنجاح من البنك الدولي ---")
print(df.head())

# 6. رسم بياني تفاعلي لمقارنة معدلات النمو الاقتصادي
fig = px.line(df, x='Year', y='GDP_Growth', color='Country',
              title='Real GDP Growth Rate Comparison (World Bank Data)',
              labels={'GDP_Growth': 'GDP Growth (%)', 'Year': 'Year'})
fig.show()
```
